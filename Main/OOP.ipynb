```python
class Cat:
    def __init__(self, name, age, colour, owner):
        self.name = name
        self.age = age
        self.colour = colour
        self.owner = owner

    def describe(self):
        print(f"{self.name} is a {self.colour} cat. "
              f"{self.name} is {self.age} years old and is owned by {self.owner}.")


cat1 = Cat("Cookie", 3, "ginger", "Steve")

cat2 = Cat("Luna", 2, "black", "Aisha")
cat3 = Cat("Milo", 5, "white", "Ben")

cat1.describe()
cat2.describe()
cat3.describe()
```

```python
class Publication:
    def __init__(self):
        self.title = ""
        self.price = 0.0

    def setdata(self):
        self.title = input()
        self.price = float(input())

    def getdata(self):
        print(self.title)
        print(self.price)

class Book(Publication):
    def __init__(self):
        super().__init__()
        self.page_count = 0

    def setdata(self):
        super().setdata()
        self.page_count = int(input())

    def getdata(self):
        super().getdata()
        print(self.page_count)

class Tape(Publication):
    def __init__(self):
        super().__init__()
        self.playing_time = 0.0

    def setdata(self):
        super().setdata()
        self.playing_time = float(input())

    def getdata(self):
        super().getdata()
        print(self.playing_time)

def main():
    book = Book()
    book.setdata()
    book.getdata()

    tape = Tape()
    tape.setdata()
    tape.getdata()

if __name__ == "__main__":
    main()
```

```python
class Publication:
    def __init__(self):
        self.title = ""
        self.price = 0.0

    def setdata(self):
        self.title = input()
        self.price = float(input())

    def getdata(self):
        print(self.title)
        print(self.price)

class Sales:
    def __init__(self):
        self.sales = [0.0, 0.0, 0.0]

    def setdata(self):
        for i in range(3):
            self.sales[i] = float(input())

    def getdata(self):
        for s in self.sales:
            print(s)

class Book(Publication, Sales):
    def __init__(self):
        Publication.__init__(self)
        Sales.__init__(self)
        self.page_count = 0

    def setdata(self):
        Publication.setdata(self)
        Sales.setdata(self)
        self.page_count = int(input())

    def getdata(self):
        Publication.getdata(self)
        Sales.getdata(self)
        print(self.page_count)

class Tape(Publication, Sales):
    def __init__(self):
        Publication.__init__(self)
        Sales.__init__(self)
        self.playing_time = 0.0

    def setdata(self):
        Publication.setdata(self)
        Sales.setdata(self)
        self.playing_time = float(input())

    def getdata(self):
        Publication.getdata(self)
        Sales.getdata(self)
        print(self.playing_time)

def exercise_objects():
    book = Book()
    book.setdata()
    book.getdata()

    tape = Tape()
    tape.setdata()
    tape.getdata()

if __name__ == "__main__":
    exercise_objects()
```

```python
class Person:
    def __init__(self, name, contact_details):
        self.name = name
        self.contact_details = contact_details

class Teacher(Person):
    def __init__(self, name, contact_details):
        super().__init__(name, contact_details)
        self.subjects = []

    def assign_subject(self, subject):
        if subject not in self.subjects:
            self.subjects.append(subject)

class Parent(Person):
    def __init__(self, name, contact_details):
        super().__init__(name, contact_details)

class Student(Person):
    def __init__(self, name, parent):
        super().__init__(name, parent.contact_details)
        self.parent = parent
        self.form_group = None

    def assign_form_group(self, form_group):
        self.form_group = form_group
        form_group.add_student(self)

class Subject:
    def __init__(self, name):
        self.name = name

class FormGroup:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        if student not in self.students:
            self.students.append(student)
            student.form_group = self

class Room:
    def __init__(self, room_number):
        self.room_number = room_number

class TimetableLesson:
    def __init__(self, subject, room, teacher, time):
        self.subject = subject
        self.room = room
        self.teacher = teacher
        self.time = time

class SchoolSystem:
    def __init__(self):
        self.teachers = []
        self.students = []
        self.parents = []
        self.subjects = []
        self.form_groups = []
        self.rooms = []
        self.timetable = []

    def add_teacher(self, name, contact_details):
        teacher = Teacher(name, contact_details)
        self.teachers.append(teacher)
        return teacher

    def add_parent(self, name, contact_details):
        parent = Parent(name, contact_details)
        self.parents.append(parent)
        return parent

    def add_student(self, name, parent):
        student = Student(name, parent)
        self.students.append(student)
        return student

    def add_subject(self, name):
        subject = Subject(name)
        self.subjects.append(subject)
        return subject

    def add_form_group(self, name):
        form_group = FormGroup(name)
        self.form_groups.append(form_group)
        return form_group

    def add_room(self, room_number):
        room = Room(room_number)
        self.rooms.append(room)
        return room

    def add_timetable_lesson(self, subject, room, teacher, time):
        lesson = TimetableLesson(subject, room, teacher, time)
        self.timetable.append(lesson)
        return lesson
```
